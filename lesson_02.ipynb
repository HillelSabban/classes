### Part 2: Checking Variable Types
Use the type() function to check what type a variable is:  
Code cell:

```python
while True:
    answer = input("Are you a student? (true/false): ").strip().lower()  # ✅ note the parentheses after .lower()

    if answer in ["true", "t", "yes", "y", "1"]:
        is_student = True
        break
    elif answer in ["false", "f", "no", "n", "0"]:
        is_student = False
        break
    else:
        print("❌ Please type 'true' or 'false'.")
        
if is_student:
    print("✅ You're a student.")
else:
    print("🚫 You aren't a student.")
```
